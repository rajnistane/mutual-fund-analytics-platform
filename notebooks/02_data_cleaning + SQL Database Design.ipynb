```python
import sqlite3
from sqlalchemy import create_engine
```

```python
engine = create_engine(r"sqlite:///C:\Users\ASUS\Desktop\Bluestock_MF_Capstone_Project\data\db\bluestock_mf.db")
```

```python
tables = {
    'dim_fund': '01_fund_master.csv',
    'fact_nav': '02_nav_history.csv',
    'fact_aum': '03_aum_by_fund_house.csv',
    'fact_sip_industry': '04_monthly_sip_inflows.csv',
    'fact_portfolio': '09_portfolio_holdings.csv',
    'fact_performance': '07_scheme_performance.csv',
    'fact_transactions': '08_investor_transactions.csv'
}
```

```python
for table, file in tables.items():
    df = pd.read_csv(rf"C:\Users\ASUS\Desktop\Bluestock_MF_Capstone_Project\data\processed\{file}")
    df.to_sql(table, engine, if_exists='replace', index=False)
    print(f"{table} loaded — {len(df)} rows")
```

```python
dates = pd.date_range(start='2022-01-01', end='2026-05-31', freq='D')

dim_date = pd.DataFrame({
    'date_id': range(1, len(dates) + 1),
    'date': dates,
    'year': dates.year,
    'month': dates.month,
    'quarter': dates.quarter,
    'is_weekday': (dates.dayofweek < 5).astype(int)
})

dim_date.to_sql('dim_date', engine, if_exists='replace', index=False)
print(f"dim_date loaded — {len(dim_date)} rows")
```

```python
def run_query(sql, title=""):
    if title:
        print(f"{title}")
    df = pd.read_sql(sql, engine)
    return df
```

```python
run_query(
    """
    SELECT scheme_name, fund_house, aum_crore 
    FROM fact_performance
    ORDER BY aum_crore DESC LIMIT 5
""",
    "1. Top 5 Funds by AUM",
)
```

```python
run_query(
    """
    SELECT strftime('%Y-%m', date) as month, 
           ROUND(AVG(nav), 2) as avg_nav
    FROM fact_nav
    GROUP BY month
    ORDER BY month
""",
    "2. Average NAV per Month",
)
```

```python
run_query("""
    SELECT month, sip_inflow_crore, yoy_growth_pct
    FROM fact_sip_industry
    ORDER BY month
""", "3. SIP Inflow YoY Growth")
```

```python
run_query("""
    SELECT state, 
           COUNT(*) as total_transactions,
           SUM(amount_inr) as total_amount
    FROM fact_transactions
    GROUP BY state
    ORDER BY total_amount DESC
""", "4. Transactions by State")
```

```python
run_query("""
    SELECT scheme_name, fund_house, expense_ratio_pct
    FROM dim_fund
    WHERE expense_ratio_pct < 1.0
    ORDER BY expense_ratio_pct ASC
""", "5. Funds with Expense Ratio < 1%")
```

```python
run_query("""
    SELECT scheme_name, fund_house, sharpe_ratio
    FROM fact_performance
    ORDER BY sharpe_ratio DESC 
    LIMIT 5
""", "6. Top 5 Funds by Sharpe Ratio")
```

```python
run_query("""
    SELECT age_group, ROUND(AVG(amount_inr), 2) as avg_sip
    FROM fact_transactions
    WHERE transaction_type = 'SIP'
    GROUP BY age_group
    ORDER BY avg_sip DESC
""", "7. Average SIP Amount by Age Group")
```

```python
run_query("""
    SELECT state, COUNT(*) as num_transactions
    FROM fact_transactions
    GROUP BY state
    ORDER BY num_transactions DESC 
    LIMIT 5
""", "8. Top 5 States by Transaction Volume")
```

```python
run_query("""
    SELECT scheme_name, fund_house, alpha, beta
    FROM fact_performance
    ORDER BY alpha DESC 
    LIMIT 5
""", "9. Top 5 Funds by Alpha")
```

```python
run_query("""
    SELECT transaction_type, 
           COUNT(*) as count,
           SUM(amount_inr) as total_amount
    FROM fact_transactions
    GROUP BY transaction_type
""", "10. Transaction Type Split")
```
