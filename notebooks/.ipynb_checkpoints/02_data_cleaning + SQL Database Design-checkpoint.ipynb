```python
import sqlite3
from sqlalchemy import create_engine
```

```python
engine = create_engine(r"sqlite:///C:\Users\ASUS\Desktop\Bluestock_MF_Capstone_Project\data\db\bluestock_mf.db")
```

```python
tables = {
    'dim_fund': '01_fund_master.csv',
    'fact_nav': '02_nav_history.csv',
    'fact_aum': '03_aum_by_fund_house.csv',
    'fact_sip_industry': '04_monthly_sip_inflows.csv',
    'fact_portfolio': '09_portfolio_holdings.csv',
    'fact_performance': '07_scheme_performance.csv',
    'fact_transactions': '08_investor_transactions.csv'
}
```

```python
for table, file in tables.items():
    df = pd.read_csv(rf"C:\Users\ASUS\Desktop\Bluestock_MF_Capstone_Project\data\processed\{file}")
    df.to_sql(table, engine, if_exists='replace', index=False)
    print(f"{table} loaded — {len(df)} rows")
```
